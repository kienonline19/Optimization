```python
import numpy as np
import matplotlib.pyplot as plt

class Cities:
    def __init__(self, n, seed=None):
        if not (isinstance(n, int) and n > 4):
            raise Exception("n needs to be an integer larger than 4")

        if seed is not None:
            np.random.seed(seed)

        self.n = n

        x = np.random.rand(n)
        y = np.random.rand(n)

        self.x, self.y = x, y


cities = Cities(5, seed=None)
cities.x
```

```python
def init_config(cities: Cities):
    return np.random.permutation(cities.n) # route

route = init_config(cities)
```

```python
def display(cities, route):
    x, y = cities.x, cities.y

    plt.clf()
    plt.axis("equal")

    plt.plot(x, y, "o", color="black")

    plt.plot(x[route], y[route], color="red")

    comeback = [route[-1], route[0]]

    plt.plot(x[comeback], y[comeback], color="red")

    plt.show()


display(cities, route)
```
