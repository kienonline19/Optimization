```python
import numpy as np

xs = np.arange(7)

costs = (xs - 3)**2 + 2

index = np.argmin(costs)

print(xs[index])
print(costs[index])
```

```python
from scipy.optimize import minimize_scalar


result = minimize_scalar(
    lambda x: (x-3)**2 + 2,
    bounds=(0, 6),
    method="bounded"
)

print(result.x)
print(result.fun)
```
